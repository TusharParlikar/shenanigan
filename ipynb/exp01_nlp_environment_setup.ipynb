```python
import nltk, spacy, gensim, transformers

# NLTK needs its data files downloaded separately
for r in ['punkt', 'punkt_tab', 'stopwords', 'averaged_perceptron_tagger_eng']:
    nltk.download(r, quiet=True)

print('NLTK        ', nltk.__version__)
print('SpaCy       ', spacy.__version__)
print('Gensim      ', gensim.__version__)
print('Transformers', transformers.__version__)
```

## NLTK: tokenize, remove stop words, POS tag

```python
from nltk.tokenize import sent_tokenize, word_tokenize
from nltk.corpus import stopwords

text = "Natural Language Processing helps computers understand human language. Python has great NLP libraries."

print(sent_tokenize(text))           # split into sentences
tokens = word_tokenize(text)         # split into words
print(tokens)

stop = set(stopwords.words('english'))
print([w for w in tokens if w.isalpha() and w.lower() not in stop])   # drop the, is, ...

print(nltk.pos_tag(tokens))          # (word, POS tag) pairs
```

## SpaCy: one call gives lemma, POS, dependency and entities

```python
nlp = spacy.load('en_core_web_sm')
doc = nlp("Sundar Pichai is the CEO of Google, which opened an office in Pune in 2024.")

for t in doc:
    print(f'{t.text:10} lemma={t.lemma_:10} pos={t.pos_:6} dep={t.dep_}')

print()
for ent in doc.ents:                 # named entities
    print(ent.text, '->', ent.label_)
```

## Gensim: train a tiny Word2Vec model
(Corpus is tiny, so similarities only prove the library works.)

```python
from gensim.models import Word2Vec

corpus = [
    "natural language processing is a field of artificial intelligence",
    "machine learning models learn patterns from data",
    "deep learning is a subset of machine learning",
    "word embeddings represent words as dense vectors",
    "artificial intelligence includes machine learning and deep learning",
]
sentences = [s.split() for s in corpus]

model = Word2Vec(sentences, vector_size=50, window=3, min_count=1, epochs=100, seed=42)

print('Vector size:', model.wv['learning'].shape)
print(model.wv.most_similar('learning', topn=5))
```

## Transformers: pre-trained sentiment model (no training needed)

```python
from transformers import pipeline

clf = pipeline('sentiment-analysis', model='distilbert-base-uncased-finetuned-sst-2-english')

for text in ["I love NLP libraries!", "The installation was slow and frustrating."]:
    print(text, '->', clf(text)[0])
```
